```python
import requests
import zipfile
import io
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
```

```python
#source: https://lvr.land.moi.gov.tw/

# outdated link
# r= requests.get('https://data.moi.gov.tw/MoiOD/System/DownloadFile.aspx?DATA=19059799-9436-4D7D-8C5E-90F1A9CE82DA', stream=True)

r = requests.get('https://plvr.land.moi.gov.tw/DownloadHistory?type=history&fileName=20251201', stream=True)

r.raise_for_status()
```

```python
with zipfile.ZipFile(io.BytesIO(r.content)) as z:
    print("files: ", z.namelist())

    target_file = 'a_lvr_land_a.csv'
    with z.open(target_file) as f_in:
        df = pd.read_csv(f_in)
```

```python
test = df.drop(columns=['都市土地使用分區', '非都市土地使用分區','非都市土地使用編定', '主要建材', 
                 '建物現況格局-房', '建物現況格局-廳', '建物現況格局-衛', '建物現況格局-隔間',
                 '有無管理組織', '車位類別', '車位移轉總面積平方公尺',
                 '編號', '電梯', '移轉編號'])
```

```python
test['交易年月日'] = pd.to_datetime(test['交易年月日'])
```

```python
test.head(2)
```

```python
test['總面積(坪)'] = np.round(np.maximum(test['土地移轉總面積平方公尺'].astype(float), test['建物移轉總面積平方公尺'].astype(float)) *  0.3025, 2)
```

```python
mask = test['建築完成年月'].notna()
test.loc[mask, '建築完成年月'] = test.loc[mask, '建築完成年月'].str[:3].astype(int).add(1911).astype(str) + test.loc[mask, '建築完成年月'].str[3:]
test['建築完成年月'] = pd.to_datetime(test['建築完成年月'])
```

```python
test[test['鄉鎮市區'] == '士林區'].sort_values(by=['交易年月日'], ascending=False).head(5)
```

```python
test.head(2)
```

```python
test['屋齡'] = test.apply(
    lambda row: 
        row['交易年月日'].year - row['建築完成年月'].year
        if (
            pd.notna(row['交易年月日']) and 
            pd.notna(row['建築完成年月']) and 
            (row['交易年月日'].year - row['建築完成年月'].year) > 0
        ) 
        else np.nan
    , axis=1
)
```

```python
test['樓別/樓高'] = test['移轉層次'] + '/' + test['總樓層數']
```

```python
test = test.drop(columns=['移轉層次', '總樓層數'])
```
